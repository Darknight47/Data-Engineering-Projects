#### Customer
1. 5,647 customers

3. 10 city codes
4. DOB approximately 1972–1992
#### Transactions
1. 23,053 rows
2. 20,878 unique transaction IDs
3. 2,057 transaction IDs have multiple rows
4. One customer per transaction ID
5. Some transaction IDs span multiple dates
6. 2,177 negative transaction lines
7. Negative Qty, Rate, and total_amt occur together
8. Financial relationship works for positive transactions
9. tran_date has inconsistent source formats
